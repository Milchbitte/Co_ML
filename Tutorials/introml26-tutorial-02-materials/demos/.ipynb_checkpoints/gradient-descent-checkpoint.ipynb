```python
%matplotlib widget
import numpy as np
import ipympl
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

def plot_gradient_descent(lambda1, lambda2, step_size, steps):
    # Define the Hessian based on eigenvalues
    H = np.array([[lambda1, 0], [0, lambda2]])
    
    # Starting point
    w = np.array([4.0, 4.0])
    history = [w.copy()]
    
    # Perform Gradient Descent
    for _ in range(steps):
        grad = H @ w
        w = w - step_size * grad
        history.append(w.copy())
    
    history = np.array(history)
    
    # Create the meshgrid for contours
    limit = 5
    x_range = np.linspace(-limit, limit, 100)
    y_range = np.linspace(-limit, limit, 100)
    X, Y = np.meshgrid(x_range, y_range)
    Z = 0.5 * (lambda1 * X**2 + lambda2 * Y**2)
    
    # Plotting
    plt.figure(figsize=(6, 6))
    plt.contour(X, Y, Z, levels=15, cmap='viridis')
    plt.plot(history[:, 0], history[:, 1], '-o', color='red', label='Path')
    plt.axhline(0, color='black', alpha=0.3)
    plt.axvline(0, color='black', alpha=0.3)
    plt.title(f"Gradient Descent (step size: {step_size:.2f})")
    plt.xlabel("$w_1$", fontsize=14)
    plt.ylabel("$w_2$", fontsize=14)
    plt.xlim(-limit, limit)
    plt.ylim(-limit, limit)
    plt.legend()
    plt.show()

# Create the interactive sliders
interact(
    plot_gradient_descent,
    lambda1=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description='λ1'),
    lambda2=FloatSlider(min=0.1, max=5.0, step=0.1, value=4.0, description='λ2'),
    step_size=FloatSlider(min=0.05, max=1.0, step=0.01, value=0.4, description='step size'),
    steps=IntSlider(min=5, max=50, step=5, value=5, description='# steps')
);
```
